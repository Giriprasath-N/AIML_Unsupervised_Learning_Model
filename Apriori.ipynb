```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_csv(
    "Market_Basket_Optimisation.csv",
    header=None
)

# Create transactions
transactions = []

for i in range(len(df)):

    transaction = []

    for item in df.iloc[i]:

        if pd.notna(item):
            transaction.append(str(item).strip())

    transactions.append(transaction)

# Encoding
te = TransactionEncoder()

encoded = te.fit(
    transactions
).transform(transactions)

basket = pd.DataFrame(
    encoded,
    columns=te.columns_
)

# Apriori
frequent = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent.head(20))

# Rules
rules = association_rules(
    frequent,
    metric="confidence",
    min_threshold=0.2
)

rules = rules.sort_values(
    "lift",
    ascending=False
)

print("\nAssociation Rules:")
print(
    rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(20)
)

# Best rule
if len(rules) > 0:

    best = rules.iloc[0]

    print("\nBest Rule:")
    print("Antecedent:", best["antecedents"])
    print("Consequent:", best["consequents"])
    print("Support:", round(best["support"], 4))
    print("Confidence:", round(best["confidence"], 4))
    print("Lift:", round(best["lift"], 4))
```
