```python
import pandas as pd
from itertools import combinations

df = pd.read_csv(
    "Market_Basket_Optimisation.csv",
    header=None
)

transactions = [
    set(str(x) for x in row if pd.notna(x))
    for _, row in df.iterrows()
]

# Vertical format
vertical = {}

for tid, transaction in enumerate(transactions):

    for item in transaction:

        if item not in vertical:
            vertical[item] = set()

        vertical[item].add(tid)

min_support = 0.02
min_count = int(
    min_support * len(transactions)
)

frequent = {
    item: tids
    for item, tids in vertical.items()
    if len(tids) >= min_count
}

results = []

items = list(frequent.keys())

for item1, item2 in combinations(items, 2):

    common = (
        frequent[item1]
        & frequent[item2]
    )

    if len(common) >= min_count:

        support = (
            len(common) /
            len(transactions)
        )

        results.append([
            item1,
            item2,
            support
        ])

result = pd.DataFrame(
    results,
    columns=["Item 1", "Item 2", "Support"]
)

result = result.sort_values(
    "Support",
    ascending=False
)

print("ECLAT")

print(result.head(10))

if len(result) > 0:
    print(
        "Best Support:",
        round(result.iloc[0]["Support"] * 100, 2),
        "%"
    )
```
